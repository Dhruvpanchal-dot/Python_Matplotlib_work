```python
import matplotlib.pyplot as plt
```

Plotting a simple graph

```python
x = [1,2,3,4,5]
y = [10,20,25,35,55]

plt.plot(x,y)
plt.show
```

```python
x = [1,2,3,4,5]
y = [10,20,25,35,55]

plt.figure(figsize=(4,2))
plt.plot(x, y, color='blue', marker='o', linestyle='--', linewidth=2, markersize=8)

plt.title("Sales Chart")
plt.xlabel("X Lable")
plt.ylabel("Y Lable")
```

Plotting graph with multiple lines

```python
x = [1,2,3,4,5]
y1 = [10,20,25,35,55]
y2 = [20,30,35,45,65]

plt.plot(x,y1, label = 'Sales of 2024')
plt.plot(x,y2, label = 'Sales of 2025')

plt.title("Year On Year Sales")
plt.xlabel("Months")
plt.ylabel("Sales")


plt.legend()
plt.show()
```

Bar Chart

```python
x=['A','B','C','D','E']
y=[25,40,50,10,20]

plt.figure(figsize=(4,2))
plt.bar(x,y)
plt.title("Bar chart")
plt.show()
```

Histogram / Used for distribution analysis

```python
import random
data = [random.randint(1, 20) for _ in range(500)]

plt.figure(figsize=(4,2))
plt.hist(data, bins=10)
plt.title("Histogram Chart")
plt.show()
```

Pie Chart

```python
Country=["India","Japan","USA","Germany","Norway"]
Sales = [20,34,57,90,50]

plt.figure(figsize=(6,4))
plt.pie(Sales , labels= Country , autopct="%1.1f%%")
plt.title("Pie Chart")
plt.show
```

Scatter Plot

```python
y1 = [10, 20, 25, 35, 45]
y2 = [20, 30, 35, 45, 55]

plt.figure(figsize=(4,2))
plt.scatter(y1, y2)
plt.title("Scatter Plot Example")
plt.show()
```

Subplot

```python
# Plotting 1st chart as Bar Chart
Country=["India","Japan","USA","Germany","Norway"]
Sales = [20,34,57,90,50]

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.bar(Country, Sales)
plt.title("Sales")
plt.xlabel("Countries")
plt.ylabel("Sales")
plt.show()

#Plotting 2nd chart as Scattered Chart
x=[10,20,30,40,50]
y=[20,30,40,50,60]

plt.subplot(1,2,2)
plt.scatter(x,y)
plt.title("Scatter Chart")
plt.show()
```

Matplotlib with Pandas 

```python
import pandas as pd

data={
    'Month':['Jan','Feb','Mar','Apr','May'],
    'Sales':[20000,30000,40000,25000,70000]
}

df=pd.DataFrame(data)
df
```

```python
plt.bar(df['Month'],df['Sales'])
plt.title("MOnthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

Simply saving any chart

```python
plt.bar(df['Month'],df['Sales'])
plt.title("MOnthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig("Monthly_sales_of_df.png")
plt.show()
```
